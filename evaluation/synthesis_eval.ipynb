# Synthesis evaluation (not production)

Offline checks for stage 3 clustering and structural claim gates.
The eight-paper file is a **smoke fixture**, not independent clustering ground truth.

Default `RUN_LIVE = False`: score a saved `synthesis_eval.json` if present; otherwise only show gold labels.

```python
from __future__ import annotations

import json
import sys
from pathlib import Path

from research_assistant.config import REPO_ROOT

if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from evaluation.synthesis_scoring import (
    adjusted_rand_index,
    citation_keep_rate,
    exclusive_labels,
    pairwise_precision_recall,
    predicted_labels,
    singleton_rate,
)

EVAL_DIR = REPO_ROOT / "evaluation"
GOLD_PATH = EVAL_DIR / "synthesis_gold.json"
SAVED_PATH = EVAL_DIR / "synthesis_eval.json"

RUN_LIVE = False

gold = json.loads(GOLD_PATH.read_text(encoding="utf-8"))
print("Gold papers:", len(gold["papers"]))
print("Notes:", gold["notes"][:120], "...")
print("RUN_LIVE:", RUN_LIVE)
```

```python
if SAVED_PATH.is_file():
    saved = json.loads(SAVED_PATH.read_text(encoding="utf-8"))
    gold_labels = exclusive_labels(gold)
    pred = predicted_labels(saved.get("assignments") or [])
    print("ARI:", adjusted_rand_index(pred, gold_labels))
    print("pairwise:", pairwise_precision_recall(pred, gold_labels))
    n = saved.get("diagnostics", {}).get("n_clustered") or len(pred)
    print("singleton_rate:", singleton_rate(saved.get("assignments") or [], n))
    print("citation_keep:", citation_keep_rate(saved.get("summaries") or [], saved.get("comparisons") or []))
else:
    print("No synthesis_eval.json yet. Offline smoke labels:")
    for row in gold["papers"]:
        print(row["arxiv_id"], row["exclusive_cluster"], row["role"])
```
